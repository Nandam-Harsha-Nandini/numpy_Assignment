## SECTION A
1.Why are NumPy arrays preferred over Python lists?
- NumPy arrays are faster than normal Python lists when working with numbers. They use memory efficiently because they store elements in a fixed data type. NumPy also provides many built-in mathematical operations, which makes calculations easier in AI/ML. Therefore, NumPy arrays are commonly used for numerical calculations.

2.Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets? 
- Vectorization means performing an operation on the whole NumPy array without using a for loop. It makes the program shorter and usually faster. This is because NumPy performs the calculations using optimized code internally. Vectorization is very useful when working with large datasets

3.For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each
- A * B means element-by-element multiplication, where corresponding elements are multiplied.
- A @ B means matrix multiplication, where rows and columns are multiplied and added.
- In neural networks, A * B can be used for element-wise operations, while A @ B is commonly used to multiply input data with weights.

4.What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise.
- Broadcasting is a NumPy feature that allows arrays of different shapes to be used in an element-wise operation. NumPy automatically adjusts the smaller array to work with the larger array when their shapes are compatible. Two dimensions are compatible if they are equal or if one of them is 1. For example, an array of shape (3, 3) can be added to an array of shape (3, 1).


5.Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why? 
- Both flatten() and ravel() are used to convert a multi-dimensional array into a one-dimensional array. flatten() creates a new copy of the array, while ravel() usually uses the original memory when possible. Therefore, in a memory-constrained application, I would prefer ravel() because it generally uses less extra memory.

6.For a 2D array, explain axis=0 and axis=1 in an aggregation function such as np.sum() or np.mean(). Illustrate with a small example. 
- For a 2D array, axis=0 means the operation is performed column-wise, while axis=1 means it is performed row-wise.
For example, for [[1, 2], [3, 4]], np.sum(array, axis=0) gives [4, 6] because 1+3=4 and 2+4=6.
np.sum(array, axis=1) gives [3, 7] because 1+2=3 and 3+4=7.

7.Why is it important to normalize or standardize numerical features before training most ML models? Explain normalization vs standardization. 
- Normalization or standardization puts numerical features into a similar range or scale. This is important because features with very large values can have more influence on some ML models. Normalization usually changes values to a fixed range such as 0 to 1, while standardization changes them so that the mean is 0 and standard deviation is 1. This can help the model train more effectively.

8.Why do functions such as np.nanmean() exist? What happens if np.mean() is used directly on data containing NaN?
- NaN means Not a Number and represents a missing or invalid numerical value. If we use np.mean() on data containing NaN, the result will usually also be NaN. np.nanmean() ignores the NaN values and calculates the mean using the available values. Therefore, it is useful when our dataset contains missing values.


9.In the ML convention X.shape = (samples, features), explain how this extends to a batch of RGB images. Write the exact shape for 64 RGB images of size 128 × 128. 
- In ML, X.shape = (samples, features) means the first value represents the number of data samples. For images, we can extend this idea by keeping the number of images first, followed by height, width, and color channels. For 64 RGB images of size 128 × 128, the exact shape is (64, 128, 128, 3). Here, 3 represents the RGB channels.


10.Why is reproducibility important during model development and experimentation? Refer to fixing a random seed with np.random.default_rng(seed). 
- Reproducibility means getting the same or very similar results when we run an experiment again. ML programs often use random values, so different runs can produce different results. We can fix the random values by using a seed, for example rng = np.random.default_rng(42). This makes experiments easier to test, compare, and debug.

```python
# D1. Data Preprocessing Pipeline
import numpy as np
data = np.array([10, 20, np.nan, 40, 50], dtype=float) # Sample 1D array containing NaN
print("Original data:")
print(data)

missing_values = np.isnan(data) # Step 1: Detect missing values
print("\nMissing value positions:")
print(missing_values)
print("Number of missing values:", np.isnan(data).sum())

mean_value = np.nanmean(data) # Step 2: Handle missing values using mean
print("\nMean value:", mean_value)
data[np.isnan(data)] = mean_value
print("Data after handling NaN:")
print(data)

mean = np.mean(data)  # Step 3: Standardize the clean data
std = np.std(data)
standardized_data = (data - mean) / std
print("\nMean after cleaning:", mean)
print("Standard deviation:", std)
print("\nStandardized data:")
print(standardized_data)
```

Explanation

Data preprocessing means cleaning and preparing data before using it in a machine learning model.
First, detect NaN values using np.isnan(), replace them with the mean using np.nanmean(), and then standardize the cleaned data using (data - mean) / std.

D3. Debug the Model Code 

There are two main bugs in the given code.
- Bug 1: Incorrect shape of w

X has shape (3, 2), meaning it has 3 rows and 2 features.
Therefore, the weight vector must contain 2 weights, one for each feature.Its shape is (3,), which does not match the 2 columns of X.
- Bug 2: Incorrect operator precedence in MSE

The given expression is:mse = np.mean(y_true - y_pred ** 2) Python calculates the power ** before subtraction. Therefore, it actually calculates:
y_true - (y_pred ** 2)
But MSE should calculate the difference first, then square it:
mse = np.mean((y_true - y_pred) ** 2)

```python
# correct code
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])
w = np.array([0.5, 1.0])
b = 2.0
y_pred = X @ w + b
y_true = np.array([10, 20, 30])
mse = np.mean((y_true - y_pred) ** 2)
print("Predicted values:", y_pred)
print("MSE:", mse)
```
